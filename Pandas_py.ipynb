```python
import pandas as pd
```

A series  is a one dimensional labelled data stucture

Pandas Series = one-dimensional labelled data structure containing values and an index

```python
marks =pd.Series([80 ,75,90,85])
```

```python
print(marks)
```

80, 75, 90, 85 → values
0, 1, 2, 3 → labels/index
It is one-dimensional → like a single column of data. we can also create our own custom labels and index

```python
marks = pd.Series ([80 ,90, 95],
index = ["john","Rahul","ramu"])
```

```python
print(marks)
```

Data Frame
A DataFrame is a two-dimensional table containing rows and columns.

```python
data = {
    "Name": [
        "John",
        "Anu",
        "Rahul"
    ],

    "Age": [
        22,
        21,
        23
    ],

    "Marks": [
        85,
        90,
        78
    ]
}

(data)
```

```python
df = pd.DataFrame(data)
df
```

isin() is used to filter rows where a column value matches any value from a given list

```python
df[
    df["Name"].isin(
        ["John", "Anu"]
    )
]
```

```python
df = df.rename(
    columns={
        "Marks": "Score"
    }
)
```
