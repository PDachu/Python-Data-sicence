linspace()   

It creates numbers that are equally spaced between a start and an end value.   

np.linspace(start, stop, num).
start → starting value.
stop → ending value
num → how many values you want

```python
import numpy as np
```

```python
np.linspace(1,10,10)
```

```python
np.linspace(0, 10,5 endpoint=True) #defult endpoint = true but we can give false on endpoint so we can stop before 10
```

```python
a = np.linspace(0, 10,5, endpoint=False)
print(a)
```
